# Training Progress Visualization

This notebook reads the training log file (`training_log.csv`) and visualizes the key metrics recorded during the training of the chess engine. The plots below show how the total loss, policy loss, and value loss change over the training steps.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import config

# Set plot style
plt.style.use('seaborn-v0_8-whitegrid')
```

## 1. Load the Training Log

First, we load the training data from the CSV log file into a pandas DataFrame. We need to make sure the log file exists.

```python
log_file = os.path.join(config.MODEL_DIR, config.TRAINING_LOG_FILE)

if not os.path.exists(log_file):
    print(f"Error: Log file not found at '{log_file}'")
    print("Please run the training script ('train.py') first to generate the log file.")
else:
    df = pd.read_csv(log_file)
    print("Training log loaded successfully.")
    display(df.head())
```

## 2. Plot the Training Losses

Now, we will plot the different loss values over the training steps. A downward trend in these plots indicates that the model is learning successfully.

```python
if 'df' in locals() and not df.empty:
    fig, axes = plt.subplots(3, 1, figsize=(12, 18), sharex=True)
    
    # Plot Total Loss
    axes[0].plot(df['step'], df['total_loss'], label='Total Loss', color='b')
    axes[0].set_ylabel('Loss')
    axes[0].set_title('Total Training Loss over Steps')
    axes[0].legend()
    
    # Plot Policy Loss
    axes[1].plot(df['step'], df['policy_loss'], label='Policy Loss', color='g')
    axes[1].set_ylabel('Loss')
    axes[1].set_title('Policy Head Loss over Steps')
    axes[1].legend()
    
    # Plot Value Loss
    axes[2].plot(df['step'], df['value_loss'], label='Value Loss', color='r')
    axes[2].set_xlabel('Training Step')
    axes[2].set_ylabel('Loss')
    axes[2].set_title('Value Head Loss over Steps')
    axes[2].legend()
    
    plt.tight_layout()
    plt.show()
else:
    print("DataFrame is not available or is empty. Cannot generate plots.")
```

### Analysis

- **Total Loss**: This is the sum of the policy and value losses. It gives an overall indication of how well the model is learning.
- **Policy Loss**: This measures how well the model is learning to predict the best move (as determined by Stockfish). A decreasing policy loss means the model is getting better at imitating Stockfish's move choices.
- **Value Loss**: This measures the difference between the model's evaluation of a position and Stockfish's evaluation. A decreasing value loss means the model's ability to evaluate positions is improving.
